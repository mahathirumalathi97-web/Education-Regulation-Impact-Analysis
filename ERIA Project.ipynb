```python
import PyPDF2

# List of PDF files (correct way)
pdf_file_paths = [
    "/content/1601817_F.No.-2-25-2024(CPP-II).pdf",
    "/content/3813291_Advisory-Safety-and-Security-of-the-Students.pdf",
    "/content/3942969_Central-Civil-Services-(Implementation-of-National-Pension-System)-Rules.pdf",
    "/content/4369607_Circular-Vigilance-Awarness-Week.pdf",
    "/content/8529252_Observance-of-Veer-Baal-Diwas-2025.pdf"
]

def extract_text_from_pdf(pdf_path):
    text = ""
    try:
        with open(pdf_path, 'rb') as file:
            reader = PyPDF2.PdfReader(file)
            for page in reader.pages:
                page_text = page.extract_text()
                if page_text:
                    text += page_text + "\n"
    except Exception as e:
        text = f"Error extracting text: {e}"
    return text

# Extract from all PDFs
all_text = ""

for path in pdf_file_paths:
    print(f"Processing: {path}")
    all_text += extract_text_from_pdf(path) + "\n\n"

print(all_text)
```

```python
def remove_whitespace(all_text):
  return  " ".join(all_text.split())
```

```python
lower_case_sentence = remove_whitespace(all_text)
print(lower_case_sentence)
```

```python
from nltk import word_tokenize # tokenized

tokenized_sentences = word_tokenize(lower_case_sentence)
print(tokenized_sentences)
```

```python
text_lower = lower_case_sentence.lower() # lower case
print(text_lower)
```

```python
import string # punctuation

text_no_punct = text_lower.translate(str.maketrans('', '', string.punctuation))
print(text_no_punct)
```

```python
from nltk.corpus import stopwords # remove stop words
import nltk
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

# Tokenize the text after punctuation removal
clean_tokens = word_tokenize(text_no_punct)

filtered_tokens = [word for word in clean_tokens if word not in stop_words]
print(filtered_tokens)
```

```python
from nltk.stem import WordNetLemmatizer # Lemmatizer
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print(lemmatized_tokens)
```
